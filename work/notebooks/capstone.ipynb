## 8. 5-minute demo outline (Week-8 showcase)

1. **Question (30s)**  
   When editors only have limited time, which pages should they review first for refresh?

2. **Method (60s)**  
   Transparent hand-rule baseline (stale + visible) vs Random Forest ranked by decline probability.  
   Validated with a client-holdout split so no client appears in both train and test.

3. **One chart / one number (60s)**  
   Precision@50: hand rule ~0.24–0.40 → model ~0.60–0.75 (roughly 2–3× on the same split).  
   Base rate ~54%.

4. **Honest result (45s)**  
   The model improves ranking at the top of the queue. It is decision-support only — it does not prove that refreshing a page recovers traffic.

5. **One recommendation (45s)**  
   Use the ranked queue with reason codes (`stale_but_visible`, `low_ctr_visible_page`, `model_decline_risk`). A human still decides the final action.
