**Target / proxy:** `is_declining_label` (binary) derived from the observed field `trend_direction == "down"`.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

The real pattern involves several interacting signals at once: freshness, position, CTR, content age, and volume.  
These relationships are non-linear and differ across clients, so a learned ranking model can surface better candidates deeper in the queue than a hand-written if-statement.
