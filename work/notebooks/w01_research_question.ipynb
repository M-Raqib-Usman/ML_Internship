## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
from pathlib import Path

# Search recursively for the anonymized dataset in the current workspace
csv_files = list(Path(".").rglob("*content_refresh_anonymized.csv"))

if not csv_files:
    # If not found, let's look one directory up as well
    csv_files = list(Path("..").rglob("*content_refresh_anonymized.csv"))

if csv_files:
    csv_path = csv_files[0]
    print(f"Found file at: {csv_path}")
    df = pd.read_csv(csv_path)

    print("Shape:", df.shape)
    print("Clients:", df["client_id"].nunique())

    down = (df["trend_direction"] == "down").sum()
    print(f"Down-trend pages: {down:,} ({100*down/len(df):.1f}%)")

    candidates = df[
        (df["trend_direction"] == "down")
        & (df["impressions_90d"] >= 500)
        & (df["avg_position"] > 0)
        & (df["avg_position"] <= 20)
    ]
    print(f"Down + ≥500 impressions + position ≤20: {len(candidates):,}")
    print(f"Mean trend_pct on those: {candidates['trend_pct'].mean():.1f}%")
else:
    print("Could not find 'content_refresh_anonymized.csv' anywhere in the current directory or its parent. Please check your file uploads or repository structure.")
```
